### **STEP 2️⃣: Check Data Quality - Missing Values** (Finding Gaps)

#### **Sub-Step 2.1: Identify Missing Values**
```python
missing_count = df.isnull().sum()
missing_percent = (missing_count / len(df)) * 100
pd.DataFrame({'Missing': missing_count, 'Percent': missing_percent})
```
**Severity levels:**
- 0% = Perfect (rare!)
- 0-2% = Ignorable (delete rows)
- 2-5% = Manageable (impute)
- 5-20% = Attention needed (strategy matters)
- 20%+ = Major issue (investigate cause)

#### **Sub-Step 2.2: Understand Missingness Pattern (MCAR vs MAR vs NMAR)**
- **MCAR** (Missing Completely At Random): Safe to delete
  - Example: Random measurement failure
  - Action: Delete rows (unbiased)
  
- **MAR** (Missing At Random): Depends on other variables
  - Example: Employees with high salary didn't provide age (embarrassed?)
  - Action: Impute using related features
  
- **NMAR** (Not Missing At Random): Value itself caused missingness
  - Example: Low salaries not reported (privacy)
  - Action: Investigate, may need special code

**How to detect:** Look for patterns. Is missingness random or systematic?

#### **Sub-Step 2.3: Visualize Missing Data**
```python
import matplotlib.pyplot as plt

plt.colorbar(label='Missing (1) vs Present (0)')
```

#### **Sub-Step 2.4: Decide Per-Column Strategy**
| Missing % | Action | Reason |
|-----------|--------|--------|
| 0% | Keep | Perfect data |
| <2% | Delete rows | Minimal loss, unbiased |
| 2-10% | Impute with median | Preserves distribution |
| 10-20% | Impute + flag | Track which rows were imputed |
| 20%+ | Investigate | Possibly drop feature if NMAR |

### **STEP 4️⃣: Detect and Handle Outliers** (Finding Anomalies)

#### **Sub-Step 4.1: Identify Outliers Using IQR Method**
```python
Q1 = df['column'].quantile(0.25)
Q3 = df['column'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = df[(df['column'] < lower) | (df['column'] > upper)]
```
**Interpretation:**
- Values beyond [Q1 - 1.5×IQR, Q3 + 1.5×IQR] are outliers
- Visualize with box plot to understand better

#### **Sub-Step 4.2: Decide: Remove, Cap, or Keep?**
- **Remove** (if data error) → df = df[~((df['col'] < lower) | (df['col'] > upper))]
- **Cap/Floor** (reduce impact) → df['col'] = df['col'].clip(lower, upper)
- **Keep** (for tree models) → Random Forest handles outliers naturally

#### **Sub-Step 4.3: Validate Decision**
```python
print(f"Before: mean={df['col'].mean():.2f}, max={df['col'].max():.0f}")

print(f"After: mean={df['col'].mean():.2f}, max={df['col'].max():.0f}")
```

# Section 1: Handling Missing Values ❌➜✅

## Learning Objectives
- Identify different types of missingness
- Choose appropriate imputation strategy
- Implement multiple imputation methods

## 📖 Theory: Types of Missing Data

**Coach's Note** 🏆  
*"Different missing patterns need different solutions!"*

1. **MCAR (Missing Completely At Random)**: Safe to delete
2. **MAR (Missing At Random)**: Use context-based imputation  
3. **NMAR (Not Missing At Random)**: Investigate the cause

## Imputation Strategies

| Method | When to Use | Pros | Cons |
|--------|-----------|------|------|
| **Delete** | <5% missing | Simple | Lose data |
| **Mean/Median** | Numerical, normal | Fast | Ignores patterns |
| **Forward Fill** | Time series | Preserves trends | Not always logical |
| **KNN** | When relationships exist | Considers similar rows | Slower |
| **Model-based** | Complex patterns | Sophisticated | Overfitting risk |

````python
print("\n" + "="*70)
print("SECTION 1: HANDLING MISSING VALUES")
print("="*70)

# Show missing data BEFORE
print("\n📊 BEFORE: Missing Values")
print(football_df[['Goals', 'Assists']].isnull().sum())
print("\nSample of data with missing values:")
print(football_df[football_df['Goals'].isnull()][['Player_Name', 'Goals', 'Assists']].head())

# APPROACH 1: Delete rows with missing values
print("\n\n✅ APPROACH 1: Drop Missing Values")
df_dropped = football_df.dropna(subset=['Goals', 'Assists'])
print(f"   Rows before: {len(football_df)} → Rows after: {len(df_dropped)}")
print(f"   ⚠️ Lost {len(football_df) - len(df_dropped)} rows!")

# APPROACH 2: Median imputation (best for outliers)
print("\n✅ APPROACH 2: Median Imputation")
df_median = football_df.copy()
df_median['Goals'].fillna(df_median['Goals'].median(), inplace=True)
df_median['Assists'].fillna(df_median['Assists'].median(), inplace=True)
print(f"   Filled missing values with median")
print(f"   Missing values now: {df_median[['Goals', 'Assists']].isnull().sum().sum()}")

# APPROACH 3: Forward fill (time series)
print("\n✅ APPROACH 3: Forward Fill (Time Series)")
df_ffill = football_df.copy()
df_ffill['Goals'] = df_ffill['Goals'].ffill()
df_ffill['Assists'] = df_ffill['Assists'].ffill()
print(f"   Filled with forward fill (previous value)")

print("\n🎯 RECOMMENDATION FOR THIS DATASET:")
print("   → Use Median Imputation (preserves distribution, handles outliers well)")
````

# Section 2: Outlier Detection & Handling 🎯

## Learning Objectives
- Detect outliers using multiple methods
- Decide when to remove vs transform outliers
- Handle outliers appropriately

## 📖 Theory: What Are Outliers?

**Coach's Note** 🏆  
*"An outlier is a data point that stands way apart from others. Find them, understand them, keep or remove them wisely!"*

**Two Types:**
- **Genuine outliers**: Real anomalies (top striker scores 50 goals)
- **Errors**: Data entry mistakes (salary = 999999)

## Detection Methods

| Method | Formula | Threshold | Use Case |
|--------|---------|-----------|----------|
| **IQR** | Q1 - 1.5×IQR to Q3 + 1.5×IQR | Beyond bounds | General purpose |
| **Z-score** | \|value - mean\| / std | Z > 3 | Assumes normal distribution |
| **Isolation Forest** | Tree-based | Anomaly score | Complex patterns |
| **Visual (Box Plot)** | Box whiskers | Visual inspection | Quick overview |

````python
print("\n" + "="*70)
print("SECTION 2: OUTLIER DETECTION & HANDLING")
print("="*70)

# Use cleaned data
df_clean = football_df.dropna(subset=['Goals', 'Assists'])

print("\n📊 STEP 1: Visualize Data Distribution")
print(f"Salary Statistics:")
print(df_clean['Weekly_Salary'].describe())

# METHOD 1: IQR Approach
print("\n✅ METHOD 1: IQR (Interquartile Range)")
Q1 = df_clean['Weekly_Salary'].quantile(0.25)
Q3 = df_clean['Weekly_Salary'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"   Q1: ${Q1:.0f}, Q3: ${Q3:.0f}")
print(f"   IQR: ${IQR:.0f}")
print(f"   Normal range: ${lower_bound:.0f} to ${upper_bound:.0f}")

outliers_iqr = df_clean[(df_clean['Weekly_Salary'] < lower_bound) | 
                        (df_clean['Weekly_Salary'] > upper_bound)]
print(f"   Outliers found: {len(outliers_iqr)}")
print(f"\n   Outlier salaries: {sorted(outliers_iqr['Weekly_Salary'].values)}")

# METHOD 2: Z-score
print("\n✅ METHOD 2: Z-Score (Standard Deviation)")
z_scores = np.abs((df_clean['Weekly_Salary'] - df_clean['Weekly_Salary'].mean()) / 
                    df_clean['Weekly_Salary'].std())
outliers_zscore = df_clean[z_scores > 3]
print(f"   Outliers (|Z| > 3): {len(outliers_zscore)} players")

# HANDLING: Three options
print("\n✅ HOW TO HANDLE OUTLIERS:")
print("\n   Option 1: Remove (if data error)")
df_removed = df_clean[~((df_clean['Weekly_Salary'] < lower_bound) | 
                        (df_clean['Weekly_Salary'] > upper_bound))]
print(f"      Rows before: {len(df_clean)} → After removal: {len(df_removed)}")

print("\n   Option 2: Cap/Floor (reduce extreme impact)")
df_capped = df_clean.copy()
df_capped['Weekly_Salary'] = df_capped['Weekly_Salary'].clip(lower=lower_bound, 
                                                              upper=upper_bound)
print(f"      Max salary after capping: ${df_capped['Weekly_Salary'].max():.0f}")

print("\n   Option 3: Keep (for tree-based models like Random Forest)")
print(f"      Trees handle outliers naturally, no need to remove!")

print("\n🎯 RECOMMENDATION:")
print("   → Cap salaries (preserve player count, reduce extreme impact)")
````

## 📖 Theory: Why Encode?

**Coach's Note** 🏆

**Two Types of Categorical Data:**

| Type | Definition | Example | Encoding |
|------|-----------|---------|----------|

````python
print("\n" + "="*70)
print("SECTION 3: FEATURE ENCODING (Categories → Numbers)")
print("="*70)

print("\n📊 BEFORE: Raw Categorical Data")
print(df_clean[['Player_Name', 'Position']].head(10))

# APPROACH 1: One-Hot Encoding (for nominal categories like Position)
print("\n✅ APPROACH 1: One-Hot Encoding (Position)")
print("   Perfect for: Nominal categories with no ordering")

df_encoded = pd.get_dummies(df_clean, columns=['Position'], prefix='Position')
print(f"\n   Original columns: Position")
print(f"   New columns: Position_Defender, Position_Goalkeeper, Position_Midfielder, Position_Striker")
print(f"\n   Result (one column per category):")
print(df_encoded[['Player_Name', 'Position_Defender', 'Position_Goalkeeper', 
                   'Position_Midfielder', 'Position_Striker']].head())

# Show why label encoding would be WRONG here
print("\n❌ WHY NOT Label Encoding (0,1,2,3)?")
position_map = {'Goalkeeper': 0, 'Defender': 1, 'Midfielder': 2, 'Striker': 3}
df_wrong = df_clean.copy()
df_wrong['Position_Label'] = df_wrong['Position'].map(position_map)
print(f"   {position_map}")
print(f"   Problem: Model thinks Striker(3) > Defender(1), but there's NO order!")
print(f"   Result: Wrong patterns learned")

print("\n🎯 FINAL ENCODING RESULT:")
print(f"   Original shape: {df_clean.shape}")
print(f"   After one-hot: ({df_encoded.shape[0]}, {df_encoded.shape[1]})")
print(f"   ✅ Position converted to {df_encoded.shape[1] - df_clean.shape[1]} binary columns")
````

# Section 4: Feature Scaling & Normalization 📏

## Learning Objectives
- Understand why scaling matters
- Apply MinMax Normalization (0 to 1)
- Apply Standardization (Z-score)
- Know when to use each

## 📖 Theory: Why Scale Features?

**Coach's Note** 🏆  
*"Different features have different ranges. A salary in millions will drown out age in decades!"*

## Two Scaling Methods

| Method | Formula | Range | Use Case | Preserves Distribution |
|--------|---------|-------|----------|----------------------|
| **MinMax (Normalization)** | (X - min) / (max - min) | [0,1] | Neural Networks | No |
| **StandardScaler (Z-score)** | (X - mean) / std | Mean=0, Std=1 | Linear models | Yes |

````python
print("\n" + "="*70)
print("SECTION 4: FEATURE SCALING & NORMALIZATION")
print("="*70)

# Select numerical columns to scale
numeric_cols = ['Age', 'Goals', 'Weekly_Salary']
df_for_scaling = df_clean[numeric_cols].copy()

print("\n📊 BEFORE Scaling (Raw Data):")
print(df_for_scaling.describe())

# METHOD 1: MinMax Scaling (0-1)
print("\n✅ METHOD 1: MinMax Normalization")
scaler_minmax = MinMaxScaler()
df_minmax = pd.DataFrame(
    scaler_minmax.fit_transform(df_for_scaling),
    columns=numeric_cols
)
print("   Range: [0, 1]")
print(df_minmax.describe())

# METHOD 2: Standard Scaling (Z-score)
print("\n✅ METHOD 2: StandardScaler (Z-score)")
scaler_standard = StandardScaler()
df_standard = pd.DataFrame(
    scaler_standard.fit_transform(df_for_scaling),
    columns=numeric_cols
)
print("   Range: Mean=0, Std=1")
print(df_standard.describe())

print("\n📊 COMPARISON (Same data, different scales):")
comparison_df = pd.DataFrame({
    'Original_Age': df_for_scaling['Age'].head(5).values,
    'MinMax_Age': df_minmax['Age'].head(5).values,
    'Standard_Age': df_standard['Age'].head(5).values
})
print(comparison_df)

print("\n🎯 WHEN TO USE EACH:")
print("   • MinMax → Neural Networks, image data, when [0,1] range is needed")
print("   • StandardScaler → Linear/Logistic Regression, data with outliers")
print("   • Tree models → No scaling needed!")
````

# Section 6: Creating Preprocessing Pipelines 🔄

## Learning Objectives
- Build reusable preprocessing pipelines
- Prevent data leakage
- Automate preprocessing steps
- Deploy pipelines to production

## 📖 Theory: Why Pipelines Matter

**Coach's Note** 🏆  
*"Manual preprocessing is error-prone. Pipelines automate and protect against mistakes!"*

## Benefits of Pipelines

✅ **Consistency**: Same steps every time  
✅ **Reproducibility**: Exactly reproduce results  
✅ **Prevention**: Protects against data leakage  
✅ **Production**: Deploy with confidence  
✅ **Maintenance**: Easy to update

````python
print("\n" + "="*70)
print("SECTION 6: PREPROCESSING PIPELINES")
print("="*70)

# Create a custom preprocessing class
class FootballPreprocessor:
    """Reusable preprocessing pipeline for football data"""
    
    def __init__(self):
        self.salary_median = None
        self.salary_bounds = None
        self.scaler = StandardScaler()
        self.encoder = None
        
    def fit(self, df):
        """Learn parameters from training data"""
        print("📚 Learning preprocessing parameters...")
        
        # Learn salary bounds
        Q1 = df['Weekly_Salary'].quantile(0.25)
        Q3 = df['Weekly_Salary'].quantile(0.75)
        IQR = Q3 - Q1
        self.salary_bounds = (Q1 - 1.5*IQR, Q3 + 1.5*IQR)
        
        # Learn scaling parameters
        self.scaler.fit(df[['Age', 'Goals']])
        
        print(f"✅ Learned salary bounds: {self.salary_bounds}")
        print(f"✅ Learned scale parameters")
        return self
    
    def transform(self, df):
        """Apply learned preprocessing"""
        df_processed = df.copy()
        
        # 1. Handle missing
        df_processed['Goals'].fillna(df_processed['Goals'].median(), inplace=True)
        
        # 2. Remove outliers
        lower, upper = self.salary_bounds
        df_processed = df_processed[(df_processed['Weekly_Salary'] >= lower) & 
                                    (df_processed['Weekly_Salary'] <= upper)]
        
        # 3. Scale features
        df_processed[['Age', 'Goals']] = self.scaler.transform(df_processed[['Age', 'Goals']])
        
        # 4. Encode categorical
        df_processed = pd.get_dummies(df_processed, columns=['Position'], drop_first=True)
        
        return df_processed

# USE THE PIPELINE
print("\n🔄 APPLYING CUSTOM PIPELINE:")
preprocessor = FootballPreprocessor()
preprocessor.fit(df_clean)

df_pipeline_output = preprocessor.transform(df_clean)
print(f"\n✅ Pipeline complete!")
print(f"   Rows: {df_clean.shape[0]} → {df_pipeline_output.shape[0]}")
print(f"   Columns: {df_clean.shape[1]} → {df_pipeline_output.shape[1]}")
print(f"\n   First few rows (processed):")
print(df_pipeline_output.head())
````

# Section 7: Preventing Data Leakage 🚫

## Learning Objectives
- Understand data leakage concept
- Identify leakage sources
- Apply proper train-test splitting
- Prevent leakage in production

## 📖 Theory: The #1 Killer of Models

**Coach's Note** 🏆  
*"Data leakage makes your model look great in testing but fails miserably in production. One of the biggest mistakes!"*

## Common Leakage Scenarios

| Leakage Type | Example | Impact |
|---|---|---|
| **Scaling Leakage** | Scale entire dataset before split | Test data influences training |
| **Target Leakage** | Use future data to predict past | Model has crystal ball |
| **Information Leakage** | Use target variable in features | Direct cheating |
| **Temporal Leakage** | Wrong time split | Look-ahead bias |

````python
print("\n" + "="*70)
print("SECTION 7: PREVENTING DATA LEAKAGE")
print("="*70)

# Simulated data for leakage demo
X = df_clean[['Age', 'Games_Played', 'Goals']].values
y = df_clean['Weekly_Salary'].values

print("\n❌ WRONG: Scale BEFORE splitting (LEAKAGE!)")
print("   1. Fit scaler on ALL data (100 players)")
scaler_wrong = StandardScaler()
X_scaled_wrong = scaler_wrong.fit_transform(X)

print("   2. Then split into train/test")
X_train_wrong, X_test_wrong, y_train, y_test = train_test_split(
    X_scaled_wrong, y, test_size=0.3, random_state=42
)
print(f"   ⚠️ PROBLEM: Scaler learned from test data!")
print(f"      Test data mean: {X_test_wrong.mean():.4f}")
print(f"      Should be ≈0 but influenced by training!")

print("\n✅ CORRECT: Split FIRST, then scale")
print("   1. Split data first")
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
print(f"   Training set: {len(X_train)} players")
print(f"   Test set: {len(X_test)} players")

print("   2. Scale learned ONLY from training data")
scaler_right = StandardScaler()
X_train_scaled = scaler_right.fit_transform(X_train)
X_test_scaled = scaler_right.transform(X_test)  # Use training parameters!

print(f"   ✅ CORRECT: Scaler learned ONLY from training!")
print(f"      Test data mean: {X_test_scaled.mean():.4f}")
print(f"      Training data mean: {X_train_scaled.mean():.4f}")

print("\n🎯 GOLDEN RULE:")
print("   Split → Preprocess → Train → Test")
print("   NEVER: Preprocess → Split → Train → Test")
````

# Section 8: Train-Test Split Strategies 📊

## Learning Objectives
- Understand different split strategies
- Use stratified splits for classification
- Handle time series data properly
- Validate models correctly

## 📖 Theory: Proper Validation

**Coach's Note** 🏆  
*"The test set is sacred. Never touch it during training!"*

````python
print("\n" + "="*70)
print("SECTION 8: TRAIN-TEST SPLIT STRATEGIES")
print("="*70)

# Prepare data for splitting
X_full = df_clean[['Age', 'Goals', 'Games_Played']]
y_full = df_clean['Weekly_Salary']

print("\n✅ STRATEGY 1: Simple Random Split (80% train, 20% test)")
X_train80, X_test20, y_train80, y_test20 = train_test_split(
    X_full, y_full, test_size=0.2, random_state=42
)
print(f"   Training: {len(X_train80)} | Test: {len(X_test20)}")

print("\n✅ STRATEGY 2: Stratified Split (preserve class distribution)")
# Create salary categories for stratification (use 2 categories to avoid too-small classes)
salary_categories = pd.qcut(y_full, q=2, labels=['Below_Median', 'Above_Median'], duplicates='drop')
X_train_strat, X_test_strat, y_train_strat, y_test_strat = train_test_split(
    X_full, y_full, test_size=0.2, stratify=salary_categories, random_state=42
)
print(f"   Training: {len(X_train_strat)} | Test: {len(X_test_strat)}")
print(f"   ✅ Class distribution preserved in both sets")

print("\n✅ STRATEGY 3: Cross-Validation (use full dataset efficiently)")
from sklearn.model_selection import KFold
kfold = KFold(n_splits=5, shuffle=True, random_state=42)
print(f"   Splitting into 5 folds for cross-validation:")
for fold_idx, (train_idx, test_idx) in enumerate(kfold.split(X_full), 1):
    print(f"   Fold {fold_idx}: Train={len(train_idx)}, Test={len(test_idx)}")

print("\n🎯 WHEN TO USE EACH:")
print("   • Simple split → Enough data, no class imbalance")
print("   • Stratified → Imbalanced datasets, classification")
print("   • Cross-val → Small datasets, maximum data usage")
````

# Section 11: Interview Questions & Answers 🎯

## Complete Interview Preparation Guide by Mohamed Waleed

### Category 1: Conceptual Understanding (Q1-Q5)

**Q1: What is data preprocessing and why is it crucial?**

**Answer**: Data preprocessing is the process of cleaning, transforming, and preparing raw data for machine learning. It's crucial because:
- Raw data contains ~20-30% errors/inconsistencies
- Poor preprocessing leads to 20-40% accuracy loss
- Takes 60-80% of data scientist's time
- "Garbage in = Garbage out"

✅ **Key Point**: Good preprocessing > Complex models

---

**Q2: What are the 6 main steps of preprocessing?**

**Answer**:
1. **Load & Explore**: Understand data structure
2. **Clean**: Handle missing values, duplicates
3. **Transform**: Scale, encode categorical
4. **Engineer**: Create new meaningful features
5. **Split**: Train-test split properly
6. **Validate**: Check quality before modeling

---

**Q3: Explain three methods to handle missing data and when to use each:**

**Answer**:
| Method | When | Pros | Cons |
|--------|------|------|------|
| **Delete** | <5% missing | Simple | Lose data |
| **Mean/Median** | Normal distribution | Fast | Loses patterns |
| **Forward Fill** | Time series | Preserves trends | Not always logical |

✅ **Interview Tip**: Ask about missingness pattern (MCAR, MAR, NMAR)

---

**Q4: What's the difference between normalization and standardization?**

**Answer**:
- **Normalization (MinMax)**: Range [0,1] using (X-min)/(max-min)
  - Use for: Neural networks, image data
  - Preserves: Distribution shape
  
- **Standardization (Z-score)**: Mean=0, Std=1 using (X-mean)/std
  - Use for: Linear models, regression
  - Assumes: Roughly normal distribution

✅ **Interview Tip**: Tree models don't need scaling!

---

**Q5: What is data leakage and give 3 examples:**

**Answer**: Data leakage = test information influences model training

Examples:
1. **Scaling leak**: Scale all data before splitting
2. **Target leak**: Use target variable in features
3. **Temporal leak**: Wrong time-based split order

✅ **Golden Rule**: Split FIRST → Preprocess SECOND

---

### Category 2: Practical Problem-Solving (Q6-Q10)

**Q6: You find 40% missing values in one column. What do you do?**

**Answer**:
1. Check if MCAR (safe to delete), MAR (use context), or NMAR (investigate)
2. If critical feature: use advanced imputation (KNN, model-based)
3. If non-critical: consider deletion
4. NEVER delete without investigation

✅ **Action**: Ask business if data is important

---

**Q7: Your model has 95% training accuracy but 60% test accuracy. What went wrong?**

**Answer**: Likely preprocessing issues:
1. **Data leakage**: Scaled before splitting
2. **Target leakage**: Used future info
3. **Feature leakage**: Used test data to create features
4. **Bad split**: Train/test distributions differ

✅ **Fix**: Review preprocessing order

**Answer**:

2. During prediction: New values become all zeros
3. Better approach: Create 'Other' category for rare values
4. Monitor: Track new values appearing

✅ **Production Ready**: Plan for unseen categories!

**Q14: How to handle imbalanced classification data?**

**Answer**:
1. **Stratified split**: Keep class ratio during train-test split
2. **Resampling**: Oversample minority or undersample majority
3. **Class weights**: Adjust model weights
4. **Metrics**: Use F1, AUC-ROC (not accuracy)

✅ **Prevention**: Check class distribution early!

**Q15: Design preprocessing for time-series data. What's different?**

**Answer**:
1. **Split differently**: Temporal order matters (no shuffling!)
2. **Lag features**: Past value as feature
3. **Rolling statistics**: Moving averages
4. **Trend removal**: Detrending for stationarity
5. **Forward fill**: For time series gaps (not random data)

✅ **Critical**: Time series ≠ random data splitting!
